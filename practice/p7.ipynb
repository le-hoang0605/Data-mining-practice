## **Q1**

Viết chương trình cài đặt 3 lớp (class) sau: `Geometry` , `Square`, và `Round`. Cho biết 2 class `Square` và `Round` kế thừa từ class `Geometry`.

Cả 3 class đều có những thuộc tính, phương thức sau:

- Thuộc tính:
    - `property` (`float`): có thể xem là độ dài cạnh (đối với hình vuông) hoặc bán kính (đối với hình tròn).

- Phương thức:
    - Hàm khởi tạo `__init__`.
    - Hàm tính diện tích `calculate_area`.

----

**Ví dụ 1**: hàm `calculate_area` của class `Square`:
- Thuộc tính của object:
```
property = 2
```
- Input:
```
# no input
```
- Output:
```
4
```

----

**Ví dụ 2**: hàm `calculate_area` của class `Round`:
- Thuộc tính của object:
```
property = 2
```
- Input:
```
# no input
```
- Output:
```
12.56637
```

**Lưu ý:** Dùng `math.pi` để tính diện tích, kết quả làm tròn 5 chữ số thập phân.

-----

##### **Yêu cầu 1:** 
Cài đặt class `Geometry`. Cài đặt phương thức `calculate_area` là không chứa implementation nào.

````python
class Geometry:
    def __init__(self, property):
        self.property=property
    def calculate_area(self):
        pass
````

````python
# Public Test
geometry_obj = Geometry(2)
assert geometry_obj.property == 2
````

##### **Yêu cầu 2:** 
Cài đặt class `Square` kế thừa từ class `Geometry`.

````python
class Square(Geometry):
    def calculate_area(self):
        return self.property ** 2
````

````python
# Public Test
square_obj = Square(2)
assert square_obj.calculate_area() == 4
````

##### **Yêu cầu 3:** 
Cài đặt class `Round` kế thừa từ class `Geometry`. Lưu ý sử dụng `math.pi` và làm tròn kết quả 5 chữ số thập phân.

````python
import math
class Round(Geometry):
    def calculate_area(self):
            return round(((self.property ** 2) * math.pi),5)
````

````python
# Public Test
square_obj = Round(2)
assert square_obj.calculate_area() == 12.56637
````

## **Q2**

Viết chương trình cài đặt lớp (class) `Fraction` (Phân số).

Lớp (class) có những thuộc tính, phương thức sau:

- Thuộc tính:
    - `numerator` (int): Tử số.
    - `denominator` (int): Mẫu số (có điều kiện khác `0`, khi khai báo sử dụng câu lệnh `assert` để kiểm tra).

- Phương thức:
    - Hàm khởi tạo `__init__`: Lưu ý trả ra lỗi nêu khai báo mẫu số bằng `0`. Lưu ý rút gọn phân số.
    - Hàm tính toán: `add`, `subtract`:
        - Cập nhật `numerator` và `denominator` của object, lưu ý rút gọn phân số.
    - Hàm rút gọn phân số: `reduce`.
    
**Quy ước:** Nếu tử số bằng `0` thì gán mẫu số bằng `1`.

- Thuộc tính của object sau khi chạy:

- Input:

other = Fraction(1, 3)

# no output

- Thuộc tính của object sau khi chạy:
```
numerator = 1
denominator = 3
```

##### **Yêu cầu:** 
Thực hiện cài đặt class `Fraction`.

````python
import math


class Fraction:
    def __init__(self, numerator, denominator):
        assert denominator != 0, "Denominator must not be zero"

        self.numerator = numerator
        self.denominator = denominator

        self.reduce()

    def reduce(self):
        if self.numerator == 0:
            self.denominator = 1
            return

        gcd = math.gcd(self.numerator, self.denominator)

        self.numerator //= gcd
        self.denominator //= gcd

        if self.denominator < 0:
            self.numerator = -self.numerator
            self.denominator = -self.denominator

    def add(self, other):
        self.numerator = (self.numerator * other.denominator + other.numerator * self.denominator)

        self.denominator = self.denominator * other.denominator

        self.reduce()

    def subtract(self, other):
        self.numerator = (self.numerator * other.denominator - other.numerator * self.denominator)

        self.denominator = self.denominator * other.denominator

        self.reduce()
````

````python
# Public Test
fraction_obj = Fraction(2, 4)
assert fraction_obj.numerator == 1
assert fraction_obj.denominator == 2
````

````python
# Public Test
try:
    has_error = False
    Fraction(1, 0)
except:
    has_error = True

assert has_error
````

````python
# Public Test
fraction_obj = Fraction(2, 3)
fraction_obj.subtract(Fraction(1, 3))
assert fraction_obj.numerator == 1
assert fraction_obj.denominator == 3
````
