### **Bài 1:** 

Một hệ thống quản lý y tế tiếp nhận một chuỗi định danh bệnh nhân thô chứa nhiều ký tự khoảng trắng, ký tự điều khiển ngắt dòng, và viết hoa không đồng nhất. Ngoài ra, mã số ID đi kèm cần được chuẩn hóa về định dạng độ dài cố định 6 ký tự bằng cách thêm các số 0 ở đầu (Zero-padding) để đồng bộ với cấu trúc cơ sở dữ liệu. Hãy xử lý chuỗi thô và tạo ra mã định danh hoàn chỉnh theo cấu trúc: `BỆNH NHÂN: [Họ Tên Chuẩn Hóa] | ID: [Mã ID Chuẩn Hóa]`.

Ví dụ 1:
- Input: 

    "  \n\t  nGuYễN vĂn a   "

    "452"
- Output:

    "BỆNH NHÂN: Nguyễn Văn A | ID: 000452"

Ví dụ 2:
- Input:

    "LÊ PHAN"

    "1234567"
- Output:

    "BỆNH NHÂN: Lê Phan | ID: 1234567"

```python
raw_name = "LÊ PHAN "
raw_id = "1234567"
# BEGIN SOLUTION
processed_name=raw_name.strip()
processed_name=processed_name.title()


processed_id=(6-len(raw_id))*str(0)+raw_id
print(processed_id)
print(processed_name)
print(f"BỆNH NHÂN: {processed_name} | ID: {processed_id}")
# END SOLUTION
```

```python
raw_name = "  \n\t  nGuYễN vĂn a   "
raw_id = "452"
# BEGIN SOLUTION
processed_name=raw_name.strip().title()

processed_id=raw_id.zfill(6)

print(processed_id)
print(processed_name)
print(f"BỆNH NHÂN: {processed_name} | ID: {processed_id}")
# END SOLUTION
```

### **Bài 4:**

Trong phân tích di truyền, một chuỗi phân đoạn DNA được biểu diễn dưới dạng: dna_seq = "ATCGGCTA". Hãy thực hiện đảo ngược hoàn toàn chuỗi ký tự trên, sau đó thay thế toàn bộ các gốc Adenine ("A") thành Uracil ("U") để tạo ra mạch phiên mã ngược RNA. Cuối cùng, tính tỷ lệ phần trăm xuất hiện của ký tự "G" trên tổng chiều dài chuỗi RNA vừa tạo.

Ví dụ:
- Input: "ATCGGCTA"
- Output: RNA: "UTCGGCUT" | Tỷ lệ G: 25.0%

```python
dna_seq = "ATCGGCTA"

# BEGIN SOLUTION
processed_dna_seq=dna_seq[::-1]
processed_dna_seq=processed_dna_seq.replace("A","U")

count_g=processed_dna_seq.count("G")
percent_g=count_g/len(processed_dna_seq)*100
# END SOLUTION
print(f"RNA: \"{processed_dna_seq}\" | Tỷ lệ G: {percent_g:.1f}%")
```

### **Bài 7:**

Trong xử lý tín hiệu y tế, kỹ thuật chuẩn hóa dữ liệu về khoảng $[0, 1]$ là bắt buộc. Cho mảng chỉ số nhiệt độ của bệnh nhân: temps = [37.5, 38.2, 39.5, 36.8, 40.0]. Hãy tính toán giá trị chuẩn hóa Min-Max cho phần tử đầu tiên và phần tử cuối cùng của mảng theo công thức: $V_{norm} = \frac{V - Min}{Max - Min}$.

Ví dụ: Với mảng temps trên $\rightarrow$ Output: Phần tử đầu normalized: 0.21875, Phần tử cuối normalized: 1.0

```python
temps = [37.5, 38.2, 39.5, 36.8, 40.0]
min_temps=min(temps)
max_temps=max(temps)

first_normalized_temps=(temps[0]-min_temps)/(max_temps-min_temps)
last_normalized_temps=(temps[-1]-min_temps)/(max_temps-min_temps)
print(f"{first_normalized_temps:.5f}")
print(f"{last_normalized_temps:.5f}")
```

### **Bài 10:**

Tọa độ không gian 3 chiều (X, Y, Z) của một trực thăng cứu thương và bệnh viện đa khoa được ghi nhận bằng hai Tuple cố định: heli_coord = (12.5, 45.0, 150.0) và hospital_coord = (15.0, 41.0, 0.0). Hãy tính toán khoảng cách không gian Euclidean thẳng giữa hai phương vị này.

Ví dụ: Với dữ liệu trên $\rightarrow$ Output: 150.07

```python
import math

heli_coord = (12.5, 45.0, 150.0)
hospital_coord = (15.0, 41.0, 0.0)

distance=math.sqrt(math.pow(heli_coord[0]-hospital_coord[0],2)+math.pow(heli_coord[1]-hospital_coord[1],2)+math.pow(heli_coord[2]-hospital_coord[2],2))

distance_theory=math.dist(hospital_coord,heli_coord)
print(f"{distance_theory:.2f}")
print(f"{distance:.2f}")
```

### **Bài 12:**

Một hệ thống tự động thu thập danh sách mã số ca nhiễm bệnh truyền nhiễm từ nhiều bệnh viện vệ tinh: raw_records = ["CV_01", "FLU_02", "CV_01", "DEN_03", "FLU_02", "CV_01"]. Hãy tính toán Hệ số dư thừa dữ liệu (Redundancy Coefficient) của hệ thống theo công thức toán học: $R = 1 - \frac{\text{Số lượng mã độc nhất}}{\text{Tổng số lượng mã thu thập thô}}$.

Ví dụ: Với raw_records trên $\rightarrow$ Output: 0.50 (Tương đương dữ liệu trùng lặp mất 50%)

```python
raw_records = ["CV_01", "FLU_02", "CV_01", "DEN_03", "FLU_02", "CV_01"]
unique_records=set(raw_records)

coefficient= 1 - (len(unique_records)/len(raw_records))
print(f"Coefficient: {coefficient:.2f}")
```

### **Bài 15:**

Trong y tế số, thuật toán chẩn đoán tự động so sánh tập hợp triệu chứng của bệnh nhân: patient_symptoms = {"Sốt", "Ho", "Khó thở"} với tập triệu chứng tiêu chuẩn của bệnh Viêm phổi: disease_symptoms = {"Sốt", "Ho", "Đau ngực", "Khó thở", "Đờm sâu"}. Hãy tính toán Chỉ số tương đồng Jaccard ($J$) theo công thức: $J = \frac{|A \cap B|}{|A \cup B|}$. Kết quả hiển thị dạng số thực làm tròn 4 chữ số thập phân.

Ví dụ: Với dữ liệu triệu chứng trên $\rightarrow$ Output: 0.6000 (Tương đồng 60%)

```python
patient_symptoms = {"Sốt", "Ho", "Khó thở"}
disease_symptoms = {"Sốt", "Ho", "Đau ngực", "Khó thở", "Đờm sâu"}

intersection_a_b=len(patient_symptoms.intersection(disease_symptoms))
union_a_b=len(patient_symptoms.union(disease_symptoms))
jaccard=intersection_a_b/union_a_b
print(jaccard)
```
