## **1. Bubble Sort**

Cho mảng n số nguyên: $a_1, a_2,..., a_n$.

Sắp xếp phần tử trong mảng bằng Bubble Sort:
- Duyệt từ đầu đến cuối danh sách, kiểm tra theo từng cặp liền kề
- Nếu thứ tự của 2 phần tử trong cặp không đúng thì đổi vị trí của 2 phẩn tử
- Tiếp tục thực hiện đến hết danh sách và đổi chỗ các cặp không đúng vị trí
- Sau khi duyệt qua toàn bộ danh sách, vị trí ở cuối danh sách là vị trí chính xác
- Lặp lại việc duyệt danh sách với danh sách mới gồm n-1 phần tử
- Lặp lại các bước trên đến khi danh sách không còn phần tử để duyệt nữa


### **Q1:**

Write a program to arrange a list in ascending order.

- Input: [23, 56, 47, 4, 19, 34, 22]
- Output: [4, 19, 22, 23, 34, 47, 56]

```python
def bubble_sort(arr: list[int]) -> list[int]:
    n = len(arr)
    for i in range(n):
        for j in range(n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr
```

### **Q5:**

After how many passes the given list will be sorted in ascending order?

- Input:
- Output:

```python
def count_passes(arr: list[int]) -> int:
    n = len(arr)
    passes = 0

    for i in range(n - 1):
        swapped = False

        for j in range(n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        passes += 1

        if not swapped:
            break

    return passes
```

## **2. Selection Sort**

Cho mảng A gồm n số nguyên: $a_1, a_2,..., a_n$.

Sắp xếp phần tử trong mảng bằng Selection Sort (in ascending order):
- Lấy giá trị đầu danh sách làm giá trị nhỏ nhất (min_value)
- Duyệt qua danh sách, so sánh min_value với các phần tử còn lại trong danh sách
- Sau khi duyệt qua danh sách, lấy phần tử có giá trị nhỏ nhất, đổi vị trí với min_value. Lúc này phần tử ở đầu danh sách đã được xếp đúng vị trí.
- Lặp lại quá trình duyệt danh sách, với danh sách bắt đầu từ phần tử thứ 2 trở đi
- Lặp lại quá trình trên đến khi danh sách không còn phần tử thì quá trình sắp xếp kết thúc

### **Q1:**

Write a program to arrange a list in ascending order.

- Input: [23, 56, 47, 4, 19, 34, 22]
- Output: [4, 19, 22, 23, 34, 47, 56]

```python
arr = [23, 56, 47, 4, 19, 34, 22]

n = len(arr)

for i in range(n - 1):
    min_index = i

    for j in range(i + 1, n):
        if arr[j] < arr[min_index]:
            min_index = j

    arr[i], arr[min_index] = arr[min_index], arr[i]

print(arr)
```

## **3. Insertion Sort**

Cho mảng A gồm n số nguyên: $a_1, a_2,..., a_n$.

Sắp xếp các phần tử trong mảng theo thứ tự tăng dần (ascending order) sử dụng Insertion Sort:
- Bắt đầu với phần tử ở vị trí thứ hai: $a_2$
- So sánh $a_2$ với $a_1$. Nếu $a_2 < a_1$, đổi chỗ 2 phần tử.
- Tiếp tục với $a_3$. So sánh $a_3$ với $a_1$ và $a_2$ và đặt $a_3$ ở đúng vị trí.
- Tiếp tục với các phần tử còn lại của mảng.

### **Q1:**

Write a program to arrange a list in ascending order.

- Input: [23, 56, 47, 4, 19, 34, 22]
- Output: [4, 19, 22, 23, 34, 47, 56]

```python
arr = list(map(int, input().split()))

for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1

    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1

    arr[j + 1] = key

print(arr)
```
