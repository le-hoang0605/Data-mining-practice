### Bài 9: Tích vô hướng 2 ma trận
Viết hàm nhận vào hai *ma trận* `X` và *ma trận* `W`, trả về giá trị tích `M` (dot product) theo công thức sau:
```
M = (x^T)W
```
Trong đó, `x^T` là phép chuyển vị của `x`. Cần kiểm tra số chiều để bảo đảm phép tính luôn tính được trước khi thực hiện phép tính, trả về giá trị `None` nếu số chiều không tương thích.

Ví dụ:


Input:
```
X = [[1, 1, 1],
    [0, 0, 0]]

W = [[1, 1, 1],
    [0, 0, 0]]
```

Output:
```
[[1, 1, 1],
 [1, 1, 1],
 [1, 1, 1]]
```

````python
def solution(X: list[list[float]], W: list[list[float]]) -> list[list[float]] | None:
    if len(X) != len(W):
        return None
    XT = []
    for j in range(len(X[0])):
        row = []
        for i in range(len(X)):
            row.append(X[i][j])
        XT.append(row)
    M = []
    for i in range(len(XT)):
        row = []
        for j in range(len(W[0])):
            value = 0
            for k in range(len(W)):
                value += XT[i][k] * W[k][j]
            row.append(value)
        M.append(row)
    return M
X = [[1, 1, 1],
    [0, 0, 0]]

W = [[1, 1, 1],
    [0, 0, 0]]
solution(X,W)
````

### Bài 10: Tổng các ma trận và vector
Viết hàm nhận vào hai ma trận `X` và `W`, trả về giá trị `M` theo công thức sau:
```
M = 2 (X^T + W).X
```
Trong đó `X^T` là phép chuyển vị của `X`. Cần kiểm tra số chiều để bảo đảm phép tính luôn tính được trước khi thực hiện phép tính, trả về giá trị `None` nếu số chiều không tương thích.

Input:
```
X = [[1, 1],
    [0, 0]
]

W = [[1, 1],
    [0, 0]
]
```

Output:
```
[[4, 4],
[2, 2]]
```

````python
def solution(X: list[list[float]], W: list[list[float]]) -> list[list[float]] | None:
    if len(X) != len(W):
        return None
    XT = []
    for j in range(len(X[0])):
        row = []
        for i in range(len(X)):
            row.append(X[i][j])
        XT.append(row)
    A = []
    for i in range(len(XT)):
        row = []
        for j in range(len(XT[0])):
            row.append(XT[i][j] + W[i][j])
        A.append(row)
    M = []
    for i in range(len(A)):
        row = []
        for j in range(len(X[0])):
            value = 0
            for k in range(len(X)):
                value += A[i][k] * X[k][j]
            row.append(2 * value)
        M.append(row)
    return M
````

### Bài 1: Gửi tiết kiệm
Tính số tiền còn lại trong tài khoản ngân hàng với đầu vào là
danh sách các số tiền giao dịch, tương ứng là sẽ có danh sách
các loại giao dịch gửi tiền (ký hiệu bằng chuỗi 'D'), rút tiền 
(ký hiệu bằng chuỗi 'W')

**Input:**
   - moneys: kiểu List, chứa danh sách các số tiền giao dịch
   (gửi và rút khỏi tài khoản)
   - types: kiểu List, chứa danh sách các loại giao dịch 
   (gửi ký hiệu là 'D', rút ký hiệu là 'W')

**Output:**
   - Kiểu số thực, số tiền còn lại trong tài khoản

Ví dụ:

Input:
```
moneys = [400.0, 100.0, 200.0]
types = ['D', 'W', 'D']
```

Output:
```
500.0
```

````python
def solution(moneys: list[float], types: list[str]) -> float:
    result = 0.0
    for i in range(len(types)):
        if types[i] == 'D':
            result += moneys[i]
        elif types[i] == 'W':
            result -= moneys[i]
    return result
````

### Bài 2: Điểm trung bình
Tính điểm trung bình của một bạn sinh viên biết trước danh sách các điểm số và số tín chỉ của từng môn học.

**Input:**
   - marks: kiểu List, gồm *n* phần tử. Đây là điểm các môn của một sinh viên.
   - credits: kiểu List, gồm *n* phần tử. Đây là số tín chỉ lần lượt của các môn.   
   
**Output:**
   - Kiểu số thực, điểm trung bình của các môn học của sinh viên

Ví dụ:

Input:
```
marks = [8.0, 9.0, 10.0]
credits = [2, 2, 1]
```

Output:
```
8.8
```

````python
def solution(marks: list[float], credits: list[int]) -> float:
    total = 0
    for i in range(len(marks)):
        total += marks[i] * credits[i]
    return total / sum(credits)
````

### Bài 3:
Đếm số lượng sinh viên có điểm số không dưới K điểm cho trước của một lớp học.

**Input:**

    - marks: kiểu List. Đây là điểm toán cả n sinh viên trong lớp.
    - K: kiểu số thực, điểm chuẩn.
   
**Output:**

    - Kiểu số nguyên, số lượng sinh viên trong lớp đạt từ K điểm trở lên

````python
def solution(marks: list[float], K: float) -> int:
    count = 0
    for mark in marks:
        if mark >= K:
            count += 1

    return count
````

# **3. Search Algorithms**

### Q1: Linear Search

Given an array, arr[] of n integers, and an integer element x, find whether element x is present in the array. Return the index of the first occurrence of x in the array, or -1 if it doesn't exist.

Input: arr[] = [1, 2, 3, 4], x = 3

Output: 2

Input: arr[] = [10, 8, 30], x = 6

Output: -1

````python
def solution(arr: list[int], x: int) -> int:
    for i in range(len(arr)):
        if arr[i] == x:
            return i

    return -1
````

### Q2: Binary Search

Given an array arr[], sorted in ascending order and an integer k. Return true if k is present in the array, otherwise, false.

Input: arr[] = [1, 2, 3, 4, 6], k = 6

Output: true

Input: arr[] = [2, 3, 5, 6], k = 1

Output: false

````python
def solution(arr: list[int], k: int) -> bool:
    left = 0
    right = len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == k:
            return True
        elif arr[mid] < k:
            left = mid + 1
        else:
            right = mid - 1
    return False
````

### Q3:

Write an efficient algorithm that searches for a value target in an m x n integer matrix matrix. This matrix has the following properties:

- Integers in each row are sorted in ascending from left to right.
- Integers in each column are sorted in ascending from top to bottom.

Input: 
```
matrix = [[1,4,7,11,15],
          [2,5,8,12,19],
          [3,6,9,16,22],
          [10,13,14,17,24],
          [18,21,23,26,30]]

target = 5
```

Output: ```true```

Input: 
```
matrix = [[1,4,7,11,15],
          [2,5,8,12,19],
          [3,6,9,16,22],
          [10,13,14,17,24],
          [18,21,23,26,30]]

target = 20
```

Output: ```false```

Constraints:
- m == matrix.length
- n == matrix[i].length
- 1 <= n, m <= 300
- -10^9 <= matrix[i][j] <= 10^9
- All the integers in each row are sorted in ascending order.
- All the integers in each column are sorted in ascending order.
- -10^9 <= target <= 10^9

````python
def solution(matrix: list[list[int]], target: int) -> bool:
    m = len(matrix)
    n = len(matrix[0])
    row = 0
    col = n - 1
    
    while row < m and col >= 0:
        current = matrix[row][col]
        if current == target:
            return True
        elif current > target:
            col -= 1
        else:
            row += 1

    return False
````
