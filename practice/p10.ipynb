### Q1:
Hiện thực hóa các phép toán đại số tuyến tính cơ bản trên không gian không gian vector $\mathbb{R}^n$ và không gian ma trận $\mathbb{R}^{m \times n}$ thông qua mảng đa chiều của NumPy.  
- Chuẩn $L_2$ (Độ dài vector): Ký hiệu là $||v||_2$, được tính bằng căn bậc hai của tổng bình phương các phần tử: $||v||=\sqrt{v_{1}^{2}+v_{2}^{2}+...+v_{n}^{2}}$.  
- Tích vô hướng (Dot Product): Biểu thị hình chiếu hình học giữa hai vector: $v\cdot u=v_{1}*u_{1}+v_{2}*u_{2}+...+v_{n}*u_{n}$.  
- Tích ma trận - vector: Biến đổi tuyến tính một vector thông qua ma trận $c = Av$.  
- Tích ma trận - ma trận: Tổng hợp hai phép biến đổi tuyến tính $C = AB$.  

Yêu cầu: Viết các hàm `compute_vector_length`, `compute_dot_product`, `matrix_multi_vector`, `matrix_multi_matrix` chỉ sử dụng thư viện numpy.

```python
import numpy as np
def compute_vector_length(vector):
    return np.sqrt(np.sum(vector ** 2))

def compute_dot_product(vector1,vector2):
    return np.dot(vector1,vector2)

def matrix_multi_vector(matrix, vector):
    return matrix @ vector

def matrix_multi_matrix(matrix1,matrix2):
    return matrix1 @ matrix2

```

### Q2: Đo lường Khoảng cách Không gian (Cosine Similarity)
Định lượng mức độ tương đồng giữa hai vector biểu diễn dữ liệu trong không gian đa chiều.  
- Độ tương đồng Cosine đánh giá góc giữa hai vector, bỏ qua độ lớn của chúng. Công thức chuẩn:
    $$cosine(x,y)=\frac{x\cdot y}{||x||||y||}=\frac{\Sigma_{1}^{n}x_{i}y_{i}}{\sqrt{\Sigma_{1}^{n}x_{i}^{2}}\sqrt{\Sigma_{1}^{n}y_{i}^{2}}}$$
- Giá trị trả về nằm trong khoảng $[-1, 1]$, trong đó $1$ thể hiện sự tương đồng tuyệt đối về hướng.  

Yêu cầu: Xây dựng hàm `compute_cosine(v1, v2)`. Tính Cosine Similarity cho hai vector: $x=[1, 2, 3, 4]$ và $y=[1, 0, 3, 0]$

```python
import numpy as np
def compute_cosine(v1,v2):
    v1_dot_v2 = np.dot(v1,v2)
    mul_v1_v2 = np.sqrt(np.sum(np.pow(v1,2))) * np.sqrt(np.sum(np.pow(v2,2)))
    return np.round((v1_dot_v2 / mul_v1_v2),5)
x=[1, 2, 3, 4]
y=[1, 0, 3, 0]

compute_cosine(x,y)
```

### Q3:
- Trích xuất các số lẻ từ mảng arr = np.arange(0, 10).  
- Thay thế toàn bộ các số lẻ trong arr bằng giá trị -1.  
- Tìm các phần tử $\geq 5$ và $\leq 10$ trong mảng [2, 6, 1, 9, 10, 3, 27].  

```python
arr = np.arange(0, 10)
new_arr = np.where(arr % 2 == 1,-1,arr)

find_arr = np.array([2, 6, 1, 9, 10, 3, 27])

index = np.where((find_arr >= 5) & (find_arr <= 10))
new_find_arr = find_arr[index]

print(new_arr)
print(new_find_arr)
```

### Q4:
Quá trình chuyển ảnh màu (RGB) sang ảnh xám được thực hiện bằng cách nén không gian 3 kênh màu theo 3 hướng tiếp cận:  
- Lightness: Trích xuất đặc trưng biên (Extrema) bằng $\frac{\max(R,G,B) + \min(R,G,B)}{2}$.  
- Average: Suy suyển thông tin tuyến tính bằng $\frac{R+G+B}{3}$.  
- Luminosity: Hiệu chỉnh trọng số theo độ nhạy quang học của mắt người bằng $0.21R + 0.72G + 0.07B$

```python
import matplotlib.image as mpimg
import numpy as np
import matplotlib.pyplot as plt

# Đọc ảnh màu thành Tensor 3D (Height, Width, Channel=3)
img = mpimg.imread("dog.jpg") 

# Phương pháp 1: Lightness
gray_lightness = (np.max(img,axis=2) + np.min(img,axis=2)) / 2

# Phương pháp 2: Average
gray_average = np.mean(img,axis=2)

# Phương pháp 3: Luminosity 
gray_luminosity = 0.21 * img[:,:,0] + 0.72 * img[:,:,1] + 0.07 * img[:,:,2]
```

```python
plt.figure(figsize=(15, 5))

# Ảnh 1: Lightness
plt.subplot(1, 3, 1) # (1 hàng, 3 cột, vị trí số 1)
plt.imshow(gray_lightness, cmap='gray')
plt.title("Method 1: Lightness")
plt.axis('off') # Tắt trục tọa độ

# Ảnh 2: Average
plt.subplot(1, 3, 2)
plt.imshow(gray_average, cmap='gray')
plt.title("Method 2: Average")
plt.axis('off')

# Ảnh 3: Luminosity
plt.subplot(1, 3, 3)
plt.imshow(gray_luminosity, cmap='gray')
plt.title("Method 3: Luminosity")
plt.axis('off')

plt.show()
```
